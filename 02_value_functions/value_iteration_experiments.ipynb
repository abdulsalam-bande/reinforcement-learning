```python
import numpy as np

# Define states and rewards
states = [0, 1, 2, 3, 4, 5]
rewards = [10, 2, 4, 6, 8, 12]
gamma = 0.9
threshold = 1e-6

# Transition probabilities
P_left = 0.2
P_right = 0.8

# Initialize value function
V = np.zeros(len(states))

# Value iteration
while True:
    delta = 0
    for s in range(len(states)):
        if s == 0 or s == 5:
            continue  # Skip terminal states
        v = V[s]
        V[s] = rewards[s] + gamma * (P_left * V[s-1] + P_right * V[s+1])
        delta = max(delta, abs(v - V[s]))
    if delta < threshold:
        break

# Extract policy
policy = []
for s in range(len(states)):
    if s == 0 or s == 5:
        policy.append("Terminal")
    else:
        left_value = P_left * V[s-1] if s > 0 else float('-inf')
        right_value = P_right * V[s+1] if s < len(states)-1 else float('-inf')
        if right_value > left_value:
            policy.append("Right")
        else:
            policy.append("Left")

# Display value function and policy
list(V), policy
```

```python
import numpy as np

# Define states and rewards
states = [0, 1, 2, 3, 4, 5]
rewards = [10, 2, 4, 6, 8, 12]
gamma = 0.9
threshold = 1e-6

# Transition probabilities
P_left = 0.2
P_right = 0.8

# Initialize value function
V = np.zeros(len(states))

# Value iteration
while True:
    delta = 0
    for s in range(1, len(states)-1):  # Skip terminal states
        v = V[s]
        V[s] = rewards[s] + gamma * (
            P_left * V[s-1] + P_right * V[s+1]
        )
        delta = max(delta, abs(v - V[s]))
    if delta < threshold:
        break

# Extract policy
policy = []
for s in range(len(states)):
    if s == 0 or s == 5:
        policy.append("Terminal")
    else:
        left_value = P_left * V[s-1] if s > 0 else float('-inf')
        right_value = P_right * V[s+1] if s < len(states)-1 else float('-inf')
        if right_value > left_value:
            policy.append("Right")
        else:
            policy.append("Left")

# Display value function and policy
list(V), policy
```
