```python
import numpy as np

# Define the number of states
n_states = 6

# Rewards vector
R = np.array([100, 0, 0, 0, 0, 40])

# Transition probability matrix for a simple scenario where non-terminal states transition to each other
P = np.zeros((n_states, n_states))
P[0, 0] = 1  # Terminal state 1
P[5, 5] = 1  # Terminal state 6

# Assuming transitions between adjacent states for simplicity
for i in range(1, 5):
    P[i, i-1] = 0.5  # Probability to go to previous state
    P[i, i+1] = 0.5  # Probability to go to next state

# Discount factor
gamma = 0.5

# Identity matrix
I = np.eye(n_states)
```

```python
P
```

```python
# Compute the value function
V = np.linalg.inv(I - gamma * P) @ R

print("Value Function:", V)
```
