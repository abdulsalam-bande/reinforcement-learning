```python
import numpy as np

# Initialize parameters
alpha = 0.01  # Learning rate
L1 = 10  # Number of samples
gamma = 0.9  # Discount factor
num_iterations = 1000  # Number of iterations

# Define the feature vector function for state-action pair
def feature_vector(s, a):
    return np.array([1, s])

# Define a simple reward function for demonstration
def reward_function(s, a):
    return s + a

# Define the transition probability function (1D MDP)
def transition(s, a):
    return s + a

# Initialize theta arbitrarily
theta = np.random.randn(2)

# Define the policy (for simplicity, use a random policy)
def policy(s):
    return np.random.choice([-1, 1])

# Trajectory Based Value Iteration (TBVI) algorithm
def tbvi(theta, alpha, L1, num_iterations):
    for _ in range(num_iterations):
        # Generate a trajectory following policy pi^e
        for s in range(-10, 11):  # Example state space
            a = policy(s)
            # Create L1 samples
            samples = [transition(s, a) for _ in range(L1)]
            # Calculate Q^+(s, a)
            Q_plus = reward_function(s, a) + gamma * max([np.dot(feature_vector(s_next, a), theta) for s_next in samples])
            print('samples',samples)
            print('1',max([np.dot(feature_vector(s_next, a), theta) for s_next in samples]))
            # Calculate Q(s, a)
            Q_sa = np.dot(feature_vector(s, a), theta)
            # Calculate delta
            delta = Q_plus - Q_sa
            # Update theta
            theta += alpha * delta * feature_vector(s, a)
    return theta

# Run the TBVI algorithm
theta = tbvi(theta, alpha, L1, num_iterations)

# Output the final theta values
print("Final theta:", theta)

# Example: Using the trained theta to evaluate a state-action pair
s_test = 5
a_test = 1
Q_value = np.dot(feature_vector(s_test, a_test), theta)
print(f"Q({s_test}, {a_test}) = {Q_value}")
```
