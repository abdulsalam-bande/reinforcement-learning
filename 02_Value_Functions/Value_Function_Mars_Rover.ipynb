```python
import random
rewards = [100, 0, 0, 0, 0, 40]
gamma = 0.5  # Discount factor
alpha = 0.1  # Learning rate
epsilon = 0.5  # Exploration rate
terminal_states = [0, 5]
num_actions = 2  # Two possible actions: 0 (left), 1 (right)
num_states = len(rewards)
number_of_episode = 10000
number_of_iterations = 1000

# Initialize Q-values to 0 for all state-action pairs
Q_values = [[0.0 for _ in range(num_actions)] for _ in range(num_states)]

def choose_action(state, epsilon):
    if random.random() < epsilon:
        return random.randint(0, num_actions - 1)  # Explore
    else:
        return max(range(num_actions), key=lambda x: Q_values[state][x])  # Exploit


for _ in range(number_of_episode):
    state = random.randint(1, num_states-1)
                           
    for _ in range(number_of_iterations):
        if state in terminal_states:
            continue
        action = choose_action(state,epsilon)
        next_state = state - 1 if action==0 else state + 1
        next_best_action_value = max(Q_values[next_state])
        Q_values[state][action] +=  alpha * (rewards[next_state] + gamma * next_best_action_value - Q_values[state][action])
        state = next_state
```
