```python
import numpy as np

# Parameters
gamma = 0.9
theta = 1e-6  # small threshold for convergence

# States
states = [1, 2, 3, 4, 5, 6]

# Rewards
rewards = {1: 100, 2: 0, 3: 0, 4: 0, 5: 0, 6: 40}

# Initial Value Function
V = {s: 0 for s in states}
V[1] = 100  # Terminal state 1
V[6] = 40   # Terminal state 6

# Initial Policy (arbitrary, let's assume initially move to the next state if possible)
policy = {1: None, 2: 1, 3: 2, 4: 3, 5: 4, 6: None}  # None for terminal states

# Transition Probabilities (assuming deterministic transitions)
P = {
    2: {3: 1.0},
    3: {4: 1.0},
    4: {5: 1.0},
    5: {6: 1.0},
}

def policy_evaluation(policy, V, gamma, theta):
    while True:
        delta = 0
        for s in states:
            if s in [1, 6]:
                continue  # Skip terminal states
            v = V[s]
            next_state = policy[s]
            V[s] = rewards[s] + gamma * V[next_state]
            delta = max(delta, abs(v - V[s]))
        if delta < theta:
            break
    return V

def policy_improvement(policy, V, gamma):
    policy_stable = True
    for s in states:
        if s in [1, 6]:
            continue  # Skip terminal states
        old_action = policy[s]
        best_value = float('-inf')
        best_action = None
        for a, next_state in P[s].items():
            value = rewards[s] + gamma * V[next_state]
            if value > best_value:
                best_value = value
                best_action = next_state
        policy[s] = best_action
        if old_action != best_action:
            policy_stable = False
    return policy, policy_stable

# Policy Iteration
while True:
    V = policy_evaluation(policy, V, gamma, theta)
    policy, policy_stable = policy_improvement(policy, V, gamma)
    if policy_stable:
        break

# Display the final value function and policy
for s in states:
    print(f"V({s}) = {V[s]}")
print("Optimal Policy:")
for s in states:
    print(f"Policy({s}) -> {policy[s]}")
```
